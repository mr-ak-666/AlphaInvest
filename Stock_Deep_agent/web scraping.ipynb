```python
from bs4 import BeautifulSoup

with open('basic_html_page.html','r') as html_file:
    content = html_file.read()
    # print(content)
    soup = BeautifulSoup(content, 'lxml')
    # print(soup)
    tags = soup.find_all('p')
    for tag in tags:
        print(tag.text)
        
```

```python

import requests
from bs4 import BeautifulSoup

# URL of Moneycontrol India news section
url = "https://www.moneycontrol.com/news/india/"


response = requests.get(url).text
# print(response)
# Parse the HTML content
soup = BeautifulSoup(response, 'lxml')

# Find news containers (adjust class based on actual HTML structure)
news_blocks = soup.find('ul', id="cagetory") # or use 'section' or 'article' tags
# print(news_blocks)


news_headers = news_blocks.find_all('a')
# print(news_headers)
i=0
for news in news_headers:
    link  = news['href']
    
    if news.find('img') or 'Remove Ad' in news.text:
        continue
    
    # print(news)
    headline = news.text.strip()
    if headline:
        i+= 1
        print(f'{i}.',headline)
        print(link)

```
